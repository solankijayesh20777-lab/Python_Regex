```python
#OOPs ke core Concept

OOPs Ke 4 Major Pillars hai
_______________________________________________________________
|                                |                             |
Encapsulation               Inheritance                   polymorphism
                                 |
                            Abstraction


1. Encapsulation  
Data or methods ko ek unit mein bind karna.

2. Inheritance 
Existing class ke features ko new class main resuse karna.

3. polymorphism
Same interface / Method ka different behavior.

4. Abstraction
Unnecessary implementaion details hide karke sirf necessary functionallity expose krna 

```

```python
OOPS ke 4 main pillars

1. Encapsulation
Data aur uses releted methods ko ek until, yani class, mein organize karna aur access ko control karna.
Ex: Bank account balance ko method ke through manage karna.

2. Inheritance
Ek Class doosri class ke features ko inharit kar sakta hain
Ex : student class, person class ke feature inherite kare.

3. polimorphisam
same interface ya method name alag objects ke liye alag behavior de skata hain
Ex : Different animals ka sount() Method.

4. Abstraction
Unnecessary implementtation details ko hide karta hai or essential functionality expose karna 
Ex : Car chalane ke liye engine ki internal working jaana jaroori hai.

    
```

# First Practical Example

```python
class student:
    def __init__(self,name,marks):
     self.name = name
     self.marks = marks
    def display(self):
        print("Name:",self.name)
        print("Marks :",self.marks)
student1 = student('Jayesh',89)
student1.display()
```

```python
Code  ka explaination
class student: student name ki class define karta hai.
__init__ object create hone par initital data set karta hai.
self.name,aur self.marks = har object ka apna data store karta hain.
display() = student ki details print b karne wala method hai.
student1 = student("Jayesh",90) = student class ka object create karta hai
student.display() = object ke trough method call karta hai.
```
