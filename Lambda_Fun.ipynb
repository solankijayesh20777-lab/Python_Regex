```python
# find even number

num = [15,20,35,22,24,26,28,30,100]
res = []
for i in num:
        if i % 2 == 0:
            res.append(i)
res
```

```python
num = [15,20,35,22,24,26,28,30,100]
def even(num):
    res = []
    for i in num:
        if i % 2 == 0:
            res.append(i)

    return res
ans = even(num)
print(ans)
```

```python
# same in lambda function

num = [15,20,35,22,24,26,28,30,100]
even = list(filter(lambda i: i % 2 ==0,num))
print(even)
```

```python
num = [15,-10,-90,20,35,22,24,26,28,30,100]
positive = list(filter(lambda i: i > 0, num))
print(positive)
```

```python
num = [15,-10,-90,20,35,22,24,26,28,30,100]
neg = list(filter(lambda i: i < 0,num))
print(neg)
```

```python
num = [40,50,60,10,20,30,70,80,90,100,105,106]
even = list(filter(lambda i: i > 50 and i % 2 == 0,num))
print(even)
```

```python
# sorted()

num = [5,2,4,6,8,9,10,20]
result = sorted(num, key = lambda i: +i)
print(result)
```

```python
num = [5,2,4,6,8,9,10,20]
result = sorted(num, key = lambda i: -i)
print(result)
```

```python
# map()

num = [1,2,3,4,5,6]
res = list(map(lambda i: i*2, num))
print(res)
```
