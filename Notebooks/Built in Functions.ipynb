# Built in Functions

```python
# abs()
numbers=[-10,-20,-30,40]
result=[abs(x) for x in numbers]
print(result)
```

```python
num=25
result=abs(num)
print(result)
```

```python
# sum()
numbers=[200,400,500,800,900,1000]
result=sum(numbers)
print(result)
```

```python
# len()
numbers=[500,400,70,80,90,100,60]
result=len(numbers)
print(result)
```

```python
# apply()
numbers=[10,20,30,40,60,80]
result=[]
for i in numbers:
    result.append(i+5)
print(result)
```

```python
# print()
numbers=[10,20,30,40,60,80]
print(numbers)
```

```python
# min()
num=[10,50,40,80,90,5]
print(min(num))
```

```python
# max()
num=[10,50,90,80,100000,900000,10000000]
print(max(num))
```

```python
# pow()
print(pow(2,3))
```

```python
# range()
for i in range(5):
    print(i)
```

```python
# Sorted()
numbers=[10,20,50,80,90,60,30,40]
print(sorted(numbers))
```

```python
# Sorted()
numbers=[10,20,50,80,90,60,30,40]
print(sorted(numbers,reverse=True))
```

```python
# Sorted()
numbers=[10,20,50,80,90,60,30,40]
print(sorted(numbers,reverse=False))
```

```python
# Reversed()
# Sorted()
numbers=[10,20,50,80,90,60,30,40]
print(list(reversed(numbers)))
```

```python
# Enumerate()
# Index+value both if we want on that case we use this function
students=['Amit','Rahul','Sneha']
for index, name in enumerate(students):
    print(index,name)
```

```python
# zip()
names=['Amit','Rahul','Sneha']
marks=[90,85,80]

for names, marks in zip(names,marks):
    print(names,marks)
```

```python
# any()
# It will check if any single value is present in our data it will return True
values=[False, True, False]
print(any(values))
```
