# List Comprehensions

```python
# Normal code
marks=[20,40,60,50,65,80,90]
print(marks)
```

```python
# Normal code
marks=[20,40,60,50,65,80,90]
new_marks=[]
for x in marks:
    new_marks.append(x+2)
print(new_marks)
```

```python
# List Comprehension
marks=[20,40,60,50,65,80,90]
new_marks=[x+2 for x in marks]
print(new_marks)
```

```python
# Find the cubes
# Normal code
cubes=[]
for x in range(10):
    if x %2==0:
        cubes.append(x**3)
print(cubes)
```

```python
# Using List Comprehension
cubes=[x**3 for x in range(10) if x%2==0]
print(cubes)
```

```python
cubes=[x**3 for x in range(10)]
print(cubes)
```

```python
# 1 to 5 ka Square
num=[1,2,3,4,5]
square=[]
for i in num:
    square.append(i*i)
print(square)
```

```python
# List Comprehension
num=[1,2,3,4,5]
square=[i*i for i in num]
print(square)
```

```python
# find the even numbers between 1 to 50
even=[]
for i in range(1,50):
    if i %2 ==0:
        even.append(i)
print(even)
```

```python
# convert in to Uppercase
# Normal code
names = ["prakash", "rahul", "amit", "rohit"]

upper_names = []

for name in names:
    upper_names.append(name.upper())

print(upper_names)
```

```python
# List Comprehension
names = ["prakash", "rahul", "amit", "rohit"]

upper_names = [name.upper() for name in names]

print(upper_names)
```

# Nested For Loop

```python
# A Nested for loop is a loop written inside for loop
for i in range(10):
    for j in range(15):
        print(i, j)
```

```python
for i in range(5):
    for j in range(5):
        print('*',end=' ')
    print()
```

```python
[[i*j for j in range(1,11)] for i in range(13,18)]
```

```python
new_list=[i for i in range(11) if i%2==0]
print(new_list)
```

```python
# Nested Condition
numbers=[1,2,3,4,5,6,18,20]
number_list=[number for number in numbers if number %2==0 if number %3==0]
print(number_list)
```

```python
# Make a list o numbers common to both the list using list comprehension
list_1=[102,132,233,234,456,564,234]
list_2=[102,233,324,564,567,523,456]
common_num=[a for a in list_1 for b in list_2 if a==b]
print(common_num)
```

```python
# Normal Code
list_1=[102,132,233,234,456,564,234]
list_2=[102,233,324,564,567,523,456]
common_num=[]
for a in list_1:
    for b in list_2:
        if a ==b:
            common_num.append(a)
print(common_num)
```
