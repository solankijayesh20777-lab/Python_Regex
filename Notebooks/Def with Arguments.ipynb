# Parameter Passing in Python

```python
def display(name):
    print('Hello',name)
display('Jayesh')
```

```python
# Positional Argument
def student(name,age):
    print('Nmae:',name)
    print('Age:',age)
student('Nitesh',21)
```

```python
# Keyword Arguments
def student(name,age):
    print('Nmae:',name)
    print('Age:',age)
student(age=21,name='Nitesh')
```

```python
# Default Arguments
def students(name,age=21):
    print('Name:',name)
    print('Age:',age)
students('Nitesh')
```

```python
# Default Arguments
def students(name,age=21):
    print('Name:',name)
    print('Age:',age)
students('Nitesh')
```

```python
# Variable-Length Arguments
# 2 Types
# 1. *args
# 2. **kwargs
```

```python
def add(*numbers):
    print(numbers)
add(10,20,30,40)
```

```python
# * args
def add(*numbers):
    total=0
    for num in numbers:
        total=total+num
    print('Total',total)
add(10,20,30,40)
```

```python
# **Kwargs
def student(**details):
    print(details)
student(name='Vyom',age=21,City='AHM',Marks=90)
```

```python
def student(**details):
    for key, value in details.items():   # items() will return the key and value
        print(key,':',value)
student(
    name='Vyom',
    age=21,
    City='AHM',
    Marks=90
)
```

# Pass by Object Reference

```python
# Arguments are passed by object reference(Object Sharing).
def change(a):
    a=100
b=500
change(b)
print(b)
```
