```python
# find even   umbers
numbers=[15,20,35,22,24,26,28,30,100]
result=[]
for i in numbers:
    if i %2 ==0:
        result.append(i)
print(result)
```

```python
numbers=[15,20,35,22,24,26,28,30,100]
def even(numbers):
    result=[]

    for i in numbers:
        if i % 2 ==0:
            result.append(i)

    return result
ans=even(numbers)
print(ans)
```

```python
# same in Lambda
# Filter()
numbers=[15,20,22,24,26,28,30,100]
even=list(filter(lambda i: i%2 ==0,numbers))
print('Even Numbers',even)
```

```python
numbers=[40,50,60,10,20,30,70,80,90,100,105,106]
even=list(filter(lambda i: i >50 and i % 2 ==0,numbers))
print('Even Numbers Greater than 50:',even)
```

```python
# Sorted()
numbers=[5,2,4,6,8,9,10,20]
result=sorted(numbers,key=lambda i:+i)
print(result)
```

```python
numbers=[5,2,4,6,8,9,10,20]
result=sorted(numbers,key=lambda i:-i)
print(result)
```

```python
# map()
numbers=[1,2,3,4,5,6]
result=list(map(lambda i: i*2, numbers))
print(result)
```
