```python
# Add Items
# Use add() method to add one item to a set
my_set={'Python','MySQL','EDA','Statistics'}
my_set.add('Machine Learning')
print(my_set)
```

```python
# Use Update() method to add more than one item
my_set=={'Python','MySQL','EDA','Statistics'}
# Update the set
my_set.update(['Machine Learning','Data Science'])
print(my_set)
```

```python
# Lenght of set
my_set={'Python','MySQl','EDA','Statistics'}
len(my_set)
```

```python
# remove() method
# Use to remove() method to remove item
my_set={'Python','EDA','Statistics','Machine Learning'}
# remove element from the set
my_set.remove('EDA')
my_set.remove('Python')
my_set
```

```python
# Use Loop to remove more than one element
my_set={'Python','MySQL','EDA','Statistics','Machine Learning'}
remove_items={'Python','EDA'}
for i in remove_items:
    my_set.remove(i)
print(my_set)
```

```python
# discard() method
my_set={'Pthn','MySQL','EDA','Statistics','Machine Learning'}
my_set.discard('Tableau')
```

```python
# remove() method
my_set={'Pthn','MySQL','EDA','Statistics','Machine Learning'}
my_set.remove('Tableau')
```

```python
# Set Operations
# Checking for subset
# Use issubset() method to check whether all elements a set are contained in anoter set
a=set([1,2,3,4,5])
b=set([1,2,3,4,5,6])
print('is a subset of b',a.issubset(b))
```

```python
a={1,2,3}
b={1,2,3,4,5}
print(b.issuperset(a))
```

```python
# Set Union()
a=set([1,2,3])
b=set([3,4,5,6])
print(a.union(b))
```

```python
# Set Intersaction
a=set([10,20,25,35,45,55,65,75,85])
b=set([10,95,105,85,115,125,135,145])
c=set([10,95,105,115,125,135,145,85])
print(a&b)
print(a&b&c)
```

```python
# Set Difference
# It returns a new set cntaining all items from the set that are not in the other set
a=set([22,45,35,85])
b=set([22,45,89,95])
print(a-b)
```
