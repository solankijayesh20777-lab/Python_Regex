# why do we use numpy

```python
Suppse we have 100000 student marks and we want to increase every student marks by 2 using normal python list, we generally need a loop.
using numpy, we can perform this oprations directly on the complete array.
```

```python
#Example

marks = [80,85,90,95,70,78,75]
n_list = []

for i in marks:
    n_list.append(i+2)
print(n_list)
```

```python
# using numpy code



marks =np.array( [80,85,90,95,70,78,75])
n_m = marks + 2
print(n_m)
```

```python
n_m = marks + 2
Numpy automatically adds 2 to every element
    we do not need to write a loop
    This feature is called vactorization.
```

```python
whay do we use array splitting?
Suppose we have 1000 students marks.
we want to divide them into:
first 300 students.
remaining 400 students.
we can use array splitting.
```

```python
# np.Split()
# The np.split() function divides an array into multiple equal-sized subquerry based on specified split position or sections.
```

```python
# Syntax

np.split(array,sections)
```

```python
# Example

# split 1D Array
out = np.split(stu_id,5)
print(out)
```

```python
#Explaination :

We divided 5 elements into 5 equal parts.
```

```python
out = np.split(stu_id,5)
for i in out:
    print(i)
```

```python
# split 2D Array

out = np.split(stu_marks,5,axis = 0)
for i in out:
    print(i)
```

```python
# Each resulating array contains one students's marks
```

```python
# splite 2D Array column wise

out = np.split(stu_marks,3,axis = 1)
for i in out:
    print(i)
```

```python
# Explaition 

# First array= python marks.
# second array = sql marks.
# Thired array = java marks
```

```python
# np.array_split()
# The np.array_split() function divides an array into a specified number of subarrays, even when the array cannot be divided into equal-size parts.
```

```python
array = np.array([101,102,103,104,105])
out = np.array_split(array,3)
for i in out:
    print(i)
```

```python
# Explation

5 Elemetns ko 3 parts me equally divide nahi kar sakte.
isiliye numpy ne :
first array = 2 elements
second array = 2 elements
thired array = 1 elements
```

```python
Diffrence between split() and array_split()

split()                                                                                  array_split()
Equal-sized splitting required for integer sections                                      unequal-sized splitting is allowed
Raises Valueerror if equal splitting is impossible                                       Divideds as evenly as possible
Used for equal divisions                                                                 Used for flexible divisions
```

```python
# Inerating a 1D Array
# One-Dimensional array interaction accesses each elements of a single-deimesional array sequentially.
```

```python
# Example ;
for i in stu_id:
    print(i)
```

```python
# Explaition :

for loop array ke elements ko one access karta hai.
i current elements ko represent karta hai.
print(i) current elements display karta hai.
```

```python
# iterating using index
# index-based interation accesses array elements using thire index positions.
```

```python
# Example

for i in range(len(stu_id)):
    print(i,stu_id[i])
```

```python
# Explaination :

len(stu_id) return 5.
range(5) generate index positions o to 4.
stu_id[i0] access each student name.
```

```python
# Interating a 2D Array
# two-dimensional array iteration accesses elements row by row elements by elements
```

```python
# Example

# rows wise interation 
for i in stu_marks:
    print(i)
```

```python
# Explaination :

# Every iteration return one complete row.
```

```python
# Elements wise iteraion

for i in stu_marks:
    for marks in i:
        print(marks)
```

```python
# Outer loop accesses each row
Inner loop accesses each elements inside that row
```

```python
# np.nditer()
# The np.nditer() function Provides an efficient way to interate over every elements of a numpy array, including
# multidimensional arrays.
```

```python
# Syntax :
# np.nditer(array) complete arrays ko iterate karta hai our har element one return karta hai.
```

```python
# Example
# Convert Elements During Interation

for i in np.nditer(stu_marks, flags = ['buffered'], op_dtypes=['float64']):
    print(i)
```

```python
# np.ndenumerate()
# The np.ndnumerate() functions through an array and returns both the index position and the corresponding elements.
```

```python
# Example

for index, value in np.ndenumerate(stu_marks):
    print(index,value)
```

```python
# Explaintion

index represents the elements positions.
values represents the actual elements.
```
