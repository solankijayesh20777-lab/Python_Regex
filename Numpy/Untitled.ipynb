# why do we use numpy

```python
Suppse we have 100000 student marks and we want to increase every student marks by 2 using normal python list, we generally need a loop.
using numpy, we can perform this oprations directly on the complete array.
```

```python
#Example

marks = [80,85,90,95,70,78,75]
n_list = []

for i in marks:
    n_list.append(i+2)
print(n_list)
```

```python
# using numpy code



marks =np.array( [80,85,90,95,70,78,75])
n_m = marks + 2
print(n_m)
```

```python
n_m = marks + 2
Numpy automatically adds 2 to every element
    we do not need to write a loop
    This feature is called vactorization.
```
