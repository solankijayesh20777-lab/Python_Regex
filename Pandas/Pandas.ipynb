# Pandas series

Waht is pandas series ?
A panda series is a one-dimensional labeled array capable of storing data of defferent data type, althogh it typically contains values of a single data type.
    Each Element in a series is accociated with an index lable.

```python
# what is a series ?

# Suppose weh have studetn marks:

import pandas as pd
marks = [85,90,78,95,65]
out =pd.Series(marks)
print(out)
```

```python
# Create the numerical series

num = range(1,100,5)
out = pd.Series(num)
print(out)
```

```python
# The output also gives the data type of the series as int64
```

```python
# To create a Series from a dictionary
# declare a dictonary

data = {'Maths':60,'Science':89,'English':76,'Social science':86}

# Create a series from the dictionary
# the dictionary keys are the index names
# the dictionary values are the Series values

pd.Series(data)
```
