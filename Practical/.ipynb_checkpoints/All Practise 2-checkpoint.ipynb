1. Type Conversion & Input Parsing   
Write a program that takes a comma-separated string of mixed floating-point numbers from the 
user, converts them into integers, calculates their sum, and prints both the list of converted integers 
and the total sum. 

```python
num = input("Enter Number => ")
list1 = num.split(",")
list2 = [int(float(i)) for i in list1]

total = sum(list2)

print(list2)
print(total)
```

2. Operator Logic & Precision Handling   
Write a function calculate_compound_interest(principal, rate, time, n) that calculates compound 
interest using the formula $A = P \left(1 + \frac{r}{n}\right)^{nt}$. Round the final result to two 
decimal places without using external modules.

```python
def calculate_compound_interest(principal,rate,time,n):
    return (principal * rate*time) / 100

calculate_compound_interest(15000,6,2,1)
    
```

3. List Modification & Element Swapping   
Given a list of numbers, write a program that swaps the first and last elements, as well as the 
second and second-to-last elements, returning the modified list.

```python
def swaping (l1 =[10,54,78,96,78,12]):
    if len(l1)>=4:
        l1[0],l1[-1] = l1[-1],l1[0]
        l1[1],l1[-2] = l1[-2],l1[1]

    return l1

swaping()
```
