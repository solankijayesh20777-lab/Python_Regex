```python
def prime(num):
    if num < 2:
        return False
    for i in range(2,num):
        if num % i == 0:
            return False
        return True
prime(12)
```

use the function created in q4 and create a dictiory where keys are prime and non prime and values are list of prime number and non prime number respectively in given range

```python
l = int(input("Enter Lower Number => "))
u = int(input("Enter Upper Number => "))

dic = {
    "prime":[w for w in range(1,u+1) if prime(w)],
    "non-prime":[w for w in range(l,u+1) if not prime(w)]
}

print(dic)
```

```python
# Q 6 use the function createdin q4 and create a dictioany where keys are prime and non prime values are count of prime numbers and non prime numbers respectively


l = int(input("Enter Lower Number => "))
u = int(input("Enter Upper Number => "))

dic = {
    "prime":len ([w for w in range(1,u+1) if prime(w)]),
    "non-prime":len([w for w in range(l,u+1) if not prime(w)])
}

print(dic)
```
