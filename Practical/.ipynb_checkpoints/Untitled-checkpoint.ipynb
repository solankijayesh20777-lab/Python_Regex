```python
def even_fibonaci(limit):
    a,b = 0,1

    while a < limit:
        if a % 2 == 0:
            yield a    
        a,b=b,a+b

for i in even_fibonaci(50):
    print(i)
```

```python
# 4. Reusing Iterators vs. Generators
# Explain what happens when you execute the following snippet. What will be printed by the second

gen = (x * x for x in range(3))

for a in gen:
    print(a, end=" ")

for b in gen:
    print(b, end=" ")
```
