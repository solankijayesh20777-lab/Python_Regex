1. Basic Squares Mapping 
Given a list numbers = [1, 2, 3, 4, 5, 6], write a list comprehension to create a new list 
containing the square of each number. 

```python
number = [1,2,3,4,5,6]
squre =[i*i for i in number]
squre

```

2. Filter Even Numbers 
Given values = [12, 5, 8, 19, 21, 14, 30], use list comprehension to extract only the even 
numbers into a new list. 

```python
values = [12, 5, 8, 19, 21, 14, 30]
even = [i for i in values if i % 2 == 0]
even
```

3. Convert Strings to Upper Case 
Given a list of words words = ["python", "java", "c++", "ruby"], generate a list of all words 
converted to uppercase using list comprehension.

```python
words = ["python", "java", "c++", "ruby"]
upper = [i.upper() for i in words]
upper
```

4. String Length Filtering 
Given animals = ["cat", "elephant", "dog", "cheetah", "giraffe", "fox"], write a list 
comprehension to pick only words with more than 4 characters. 

```python
string = ["cat","elephant","dog","cheetah","giraffe","fox"]
ch = [i for i in string if len(i) > 4 ]
ch
```

5. If-Else Conditional Expression 
Given scores = [45, 82, 70, 38, 90], write a list comprehension that outputs "Pass" if the 
score is 50 or above, and "Fail" otherwise. 

```python
scores = [45, 82, 70, 38, 90]
mark = ["Pass" if i >= 50 else "Fail" for i in scores]
mark
```

6. Flatten a 2D Matrix 
Given a 2D list matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]], write a nested list comprehension to 
flatten it into a single 1D list [1, 2, 3, 4, 5, 6, 7, 8, 9]. 

```python
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
single = [i for j in matrix for i in matrix ]
single
```

7. Extract Specific Vowels 
Given the string sentence = "Python Programming Language", write a list comprehension 
to collect all vowels (a, e, i, o, u) present in the string (case-insensitive). 

```python
sentence = "Python Programming Language"
vowels = [i for i in sentence if i.lower() in 'aeiou']
vowels
```
