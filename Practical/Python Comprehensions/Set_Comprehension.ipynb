8. Unique Squares 
Given nums = [1, -2, 2, -3, 3, 4, -4], use set comprehension to calculate the square of each 
number to ensure all result values are unique. 

```python
nums = [1, -2, 2, -3, 3, 4, -4]
squre = {i*i for i in nums}
squre
```

9. Unique Word Lengths 
Given a list words = ["apple", "banana", "kiwi", "pear", "mango", "fig", "plum"], create a set 
containing the unique lengths of these words.

```python
words = ["apple", "banana", "kiwi", "pear", "mango", "fig", "plum"]
length = {len(i) for i in words}
length
```

10. Extract Unique First Letters 
Given names = ["Alice", "Bob", "Amanda", "Charlie", "Brian", "David"], use set 
comprehension to extract the first letter of each name in uppercase. 


```python
names = ["Alice", "Bob", "Amanda", "Charlie", "Brian", "David"]
uper = {i for i in names if i.upper()}
uper
```

11. Filter Odd Numbers into a Set 
Given data = [10, 15, 20, 25, 30, 35, 40], collect only the odd numbers into a set using set 
comprehension. 

```python
data = [10, 15, 20, 25, 30, 35, 40]
odd = {i for i in data if i % 2 != 0}
odd
```

12. Divisible by Both 3 and 5 
Generate a set of numbers between 1 and 50 (inclusive) that are evenly divisible by both 3 
and 5 using set comprehension. 

```python
divi = {i for i in range(1,50) if i % 3 == 0 and if i % 5 ==0}
print(divi)
```
