Question 10
Which keyword allows you to modify a global variable inside a function?
● A) nonlocal
● B) global
● C) import
● D) extern


ans - >  B) global

```python
x = 10

def change():
    global x
    x = 20

change()
print(x)
```

Question 16
Write a user-defined function named is_positive that takes one number as a parameter and
returns True if the number is greater than 0, otherwise False.




```python
def positive(num):
    if num > 0:
        return True
    else:
        return False

print(positive(5))
print(positive(-3))
print(positive(0))
```
