# 1. Encapsulation

```python
Encapsulation means combining data and methods inside a class and controlling access to that data.
Data + Methods + Controlls Access = Ecapsulation 
Why Use?
Data Ko Directly unwanted modification se protect/contro; karne ke liye
```

```python
# Example Of Encapulation

class BankAccount:
    def __init__(self,balance):
        self.__balance = balance

    def deposit(self,amount):
        self.__balance += amount

    def get_balance(self):
        return self.__balance

account = BankAccount(100000)
account.deposit(50000)
print(account.get_balance())
```

```python
self.__balance
# __balance ko class ke bahar directly access karna intended interface nahi hai.

# Instead:

account.deposit(50000)
se controlled modification hota hai.

# Real life Example

ATM main aap directly ban database ka balance change nahi kar sakte.

    ATM
     |
 Deposite / Withdraw
     |
 Bank System 
     |
  Balance

Iske trough hum controll opration karte he.

# One Linr memory
# Encapsulation = Data ko bundle karna + access ko control karna.

    
```

# 2. Inheritance

```python
Inheritance allows the one class to acquire the properies and methods of another class.


Exsiting class ke features ko new class main resuse karna.

why use ?
sabse importent reason :

code resubality
```

```python
# Example :

class Animal:
    def eat(self):
        print('Animal eats')

    def sleep(self):
        print('Animal sleep')
```

```python
class Dog(Animal):
    pass
```

```python
dog = Dog()
dog.eat()
dog.sleep()
```

```python
# Parent & child

    Animal
      |
    eat()
    sleep()
        |
        |inherists
        |
    Dog

# Yhan:
Animal - parent / base / superclass
Dog- child / deerived / subclass
```

```python
# Chiled open feature bhi add kar sakte hai

class Dog(Animal):
    def bark(self):
        print('Dog barks')
```

```python
dog = Dog()
dog.eat()
dog.sleep()
dog.bark()
```

```python
# Inheritance = Exisiting class ke feture ko resuse karna.
```

# 3. Polymorphism

```python
Ye OOPs ka bahut importent concept hai.

Defination.
poly = many
Morphism = Forms

Polymorphism = One interface / name, multiple forms / behavoyors. 
    
Same method name, diffrent behavior.
```

```python
# Example

class Dog:
    def sound(self):
        print('Dog barks')

class cat:
    def sound(self):
        print('cat meows')
        
```

```python
dog = Dog()
cat = cat()

dog.sound()
cat.sound()
```

```python
Notice one thing 
dog.sound()
cat.sound()

method dono ke sem hai
behaviyar deefrent hai
```

```python
Sound()
Lekin behavior deffrent hai.
```

```python
Dog -sound()-Break
Cat-sound()-meow
Yahi Polimorphism hai.
```

# Abstraction

```python
Abstraction means hiding unnecessary implimentation details and exposing only the essential functinality.

user ko 'what to do ' dikhata hai, 'How it works internally' hide karna
```

```python
Real Life Example :

Aap car drive karte ho.
aap
start
Acceleterator
brake
steering
Use Karte ho
```

```python
Lekin aapko internally ye nahi dekhna padta.
Fuel insaction
Engine Combustion
Piston movement
Transmission mechanism
ECu processing
Apko sirf required interface diya gaya hai.
 this is abstraction
```

```python
# Example

from abc import ABC,abstraction
```

```python
class Animal(ABC):
    @abstractionmethod
    def sound(self):
        pass
```

```python
# Ab child classes ko sound() implement karna padega:
class Dog(Animal):
    def sound(self):
        print("Dog Barks")

class cat(Animal):
    def sound(self):
        print("Cat meows")
```

```python
dog = Dog()
cat = cat()

dog.sound()
cat.sound()
```

```python
# Yaha Abstraction kaha hai ?

Animal bol raha hai:
def sound(self):
Har animal ka sound hona chahiye.
Lekin Animal ye nahi batata ki sound kaise produce hoga.
Dog decide karega:
print("Dog barks")

cat decide karegi :
print("Cat meows")

So :
    Animal 
      |
      | Tells what 
      |
    sound()

     Dog - How - Bark
     Cat - Show - meow

Abstraction = Importent cheez expose karo,Unnnecesarry implementation hide karo
```
